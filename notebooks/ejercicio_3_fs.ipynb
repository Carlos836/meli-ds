```python

# Feature Selection con Ensembles Débiles vs Ruido
import pandas as pd
import numpy as np
import lightgbm as lgb
import catboost as cb
import xgboost as xgb
from collections import Counter

def feature_selection_weak_ensembles(X, y, threshold=2, random_state=42):
    """
    Feature selection usando ensembles débiles vs variable de ruido.
    
    Parámetros:
    - X: DataFrame con features predictoras
    - y: Series con variable respuesta
    - threshold: número mínimo de modelos que deben superar el ruido (default=2)
    - random_state: semilla para reproducibilidad
    
    Retorna:
    - selected_features: lista de features seleccionadas
    - importance_results: DataFrame con resultados de importancia
    """
    
    # Agregar variable de ruido
    np.random.seed(random_state)
    X_with_noise = X.copy()
    X_with_noise['noise'] = np.random.normal(0, 1, len(X))
    
    print(f"Dataset original: {X.shape[1]} features")
    print(f"Dataset con ruido: {X_with_noise.shape[1]} features")
    print(f"Umbral de decisión: {threshold} modelos")
    
    # Configuración de modelos débiles (máximo 2 hiperparámetros)
    models = {
        'LightGBM': lgb.LGBMClassifier(
            n_estimators=50,   # pocos árboles
            max_depth=3,       # poca profundidad
            random_state=random_state,
            verbose=-1
        ),
        'CatBoost': cb.CatBoostClassifier(
            iterations=50,     # pocas iteraciones
            depth=3,           # poca profundidad
            random_state=random_state,
            verbose=False
        ),
        'XGBoost': xgb.XGBClassifier(
            n_estimators=50,  # pocos árboles
            max_depth=3,       # poca profundidad
            random_state=random_state,
            verbosity=0,
            missing=np.nan
        )
    }
    
    # Calcular importancia para cada modelo
    importance_results = []
    
    for model_name, model in models.items():
        print(f"\nEntrenando {model_name}...")
        
        try:
            # Entrenar modelo
            model.fit(X_with_noise, y)
            
            # Obtener importancia
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
            elif hasattr(model, 'get_feature_importance'):
                importances = model.get_feature_importance()
            else:
                print(f"⚠️ {model_name}: No se pudo obtener importancia")
                continue
            
            # Crear DataFrame con resultados
            feature_importance = pd.DataFrame({
                'feature': X_with_noise.columns,
                'importance': importances,
                'model': model_name
            })
            
            importance_results.append(feature_importance)
            
            # Mostrar top 5 features
            top_features = feature_importance.nlargest(5, 'importance')
            print(f"Top 5 features para {model_name}:")
            print(top_features[['feature', 'importance']].to_string(index=False))
            
        except Exception as e:
            print(f"❌ Error en {model_name}: {str(e)}")
            continue
    
    # Combinar resultados
    if not importance_results:
        print("❌ No se pudieron calcular importancias")
        return [], pd.DataFrame()
    
    all_results = pd.concat(importance_results, ignore_index=True)
    
    # Calcular importancia del ruido por modelo
    noise_importance = all_results[all_results['feature'] == 'noise'].copy()
    print(f"\n=== IMPORTANCIA DEL RUIDO POR MODELO ===")
    print(noise_importance[['model', 'importance']].to_string(index=False))
    
    # Identificar features que superan el ruido
    features_above_noise = {}
    
    for model_name in models.keys():
        model_results = all_results[all_results['model'] == model_name]
        
        # Verificar que existe la columna noise para este modelo
        noise_data = model_results[model_results['feature'] == 'noise']
        if len(noise_data) == 0:
            print(f"⚠️ {model_name}: No se encontró columna 'noise' - saltando")
            continue
            
        noise_imp = noise_data['importance'].iloc[0]
        
        # Features que superan el ruido
        above_noise = model_results[
            (model_results['feature'] != 'noise') & 
            (model_results['importance'] > noise_imp)
        ]['feature'].tolist()
        
        features_above_noise[model_name] = above_noise
        print(f"\n{model_name}: {len(above_noise)} features superan el ruido")
    
    # Contar cuántos modelos seleccionan cada feature
    feature_votes = Counter()
    for model_features in features_above_noise.values():
        for feature in model_features:
            feature_votes[feature] += 1
    
    # Seleccionar features que superan el umbral
    selected_features = [
        feature for feature, votes in feature_votes.items() 
        if votes >= threshold
    ]
    
    print(f"\n=== RESULTADO FINAL ===")
    print(f"Features seleccionadas (≥{threshold} modelos): {len(selected_features)}")
    print(f"Features originales: {X.shape[1]}")
    print(f"Reducción: {((X.shape[1] - len(selected_features)) / X.shape[1] * 100):.1f}%")
    
    # Mostrar features seleccionadas con sus votos
    selected_votes = [(f, feature_votes[f]) for f in selected_features]
    selected_votes.sort(key=lambda x: x[1], reverse=True)
    
    print(f"\nFeatures seleccionadas:")
    for feature, votes in selected_votes:
        print(f"  {feature}: {votes} modelos")
    
    return selected_features, all_results
```

```python
selected_features, importance_results = feature_selection_weak_ensembles(X, y, threshold=1)
```
